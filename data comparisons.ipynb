# Comparing datasets
seeing if the new test data is the issue and checking for similar correlations in the two .csv files

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
train = pd.read_csv("datasets/life-expectancy.csv")
test = pd.read_csv('datasets/life_expectancy_test_data.csv')
```

### Visualizing the Relationship of Age to Other Columns


```python
# Identify categorical and numerical columns (excluding 'Age')
categorical_cols = train.select_dtypes(include=['object', 'category', 'string']).columns.tolist()
numeric_cols = train.select_dtypes(include=[np.number]).columns.tolist()
if 'Age' in numeric_cols:
    numeric_cols.remove('Age')

import seaborn as sns

# Plot numeric columns: scatter plots
for col in numeric_cols:
    fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharey=True)
    sns.scatterplot(ax=axes[0], data=train, x=col, y='Age', alpha=0.6)
    axes[0].set_title(f'Train: Age vs {col}')
    sns.scatterplot(ax=axes[1], data=test, x=col, y='Age', alpha=0.6)
    axes[1].set_title(f'Test: Age vs {col}')
    plt.tight_layout()
    plt.show()

# Plot categorical columns: boxplots
for col in categorical_cols:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)
    sns.boxplot(ax=axes[0], data=train, x=col, y='Age')
    axes[0].set_title(f'Train: Age by {col}')
    sns.boxplot(ax=axes[1], data=test, x=col, y='Age')
    axes[1].set_title(f'Test: Age by {col}')
    plt.tight_layout()
    plt.show()
```
