```python
import pandas as pd
import numpy as np
from joblib import load
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer 
from sklearn.pipeline import Pipeline
```

```python
# Load the lr model back into a variable
loaded_model = load('lrmodel.joblib')
```

```python
#importing the test dataset cerated by my friend chatgpt
test_data = pd.read_csv('datasets/life_expectancy_test_data.csv')
test_data = test_data.sample(frac=1, random_state=1)
test_data.head()
```

```python
X_cat = test_data.select_dtypes(include=["object", "string"]).columns
X_num = test_data.select_dtypes(exclude=["object", "string"]).columns

features = test_data.keys()
cat_features_ind =[0,4,5,6,7,8,10,11,12,13]
X_cat = [features[i] for i in cat_features_ind]
print("categorical:")
print(X_cat)
num_features_ind = [1,2,3,9]
X_num = [features[i] for i in num_features_ind]
print("numerical:")
print(X_num)
```

num_transformer = Pipeline(
    steps = [
        ('imputer', SimpleImputer(strategy='mean')),
        ('scaler', StandardScaler())  
    ]
)

cat_transformer = Pipeline(
    steps = [
        ('imputer', SimpleImputer(strategy='constant', fill_value='Missing')),
        ('onehot', OneHotEncoder(sparse_output=False, handle_unknown='ignore'))
    ]
)
preprocessor = ColumnTransformer(
    transformers = [
        ('num', num_transformer, X_num),
        ('cat', cat_transformer, X_cat)
    ]
)
y = test_data["Age"]
X = preprocessor.fit_transform(test_data)
print(np.shape(X))

```python
preprocessor = load('preprocessor.joblib')
y = test_data["Age"]
X = preprocessor.transform(test_data)
print(np.shape(X))
```

```python
predictions = loaded_model.predict(X,)
```

```python
print(predictions)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
print("Mean squared error: %.2f" % mean_squared_error(y, predictions))
# The coefficient of determination: 1 is perfect prediction
print("Coefficient of determination: %.2f" % r2_score(y, predictions))
```
